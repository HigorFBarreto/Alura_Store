# 4. Produtos Mais e Menos Vendidos

```python
# Função para plotar os 5 produtos mais e os 5 menos vendidos
def produtos_mais_menos_vendidos(df, nome_loja):
    contagem_produtos = df['Produto'].value_counts()

    top_mais = contagem_produtos.head(5)
    top_menos = contagem_produtos.tail(5)

    fig, axs = plt.subplots(1, 2, figsize=(15, 5))

    # Mais vendidos
    bars1 = axs[0].bar(top_mais.index, top_mais.values, color='mediumblue')
    axs[0].set_title(f'Top 5 Mais Vendidos - {nome_loja}')
    axs[0].set_ylabel('Quantidade Vendida')
    axs[0].tick_params(axis='x', rotation=45)
    for bar in bars1:
        yval = bar.get_height()
        axs[0].text(bar.get_x() + bar.get_width()/2.0, yval + 1, int(yval), ha='center', va='bottom', fontsize=9)

    # Menos vendidos
    bars2 = axs[1].bar(top_menos.index, top_menos.values, color='tomato')
    axs[1].set_title(f'Top 5 Menos Vendidos - {nome_loja}')
    axs[1].set_ylabel('Quantidade Vendida')
    axs[1].tick_params(axis='x', rotation=45)
    for bar in bars2:
        yval = bar.get_height()
        axs[1].text(bar.get_x() + bar.get_width()/2.0, yval + 1, int(yval), ha='center', va='bottom', fontsize=9)

    plt.tight_layout()
    plt.show()

# Aplicando para cada loja
produtos_mais_menos_vendidos(loja_1, "Loja 1")
produtos_mais_menos_vendidos(loja_2, "Loja 2")
produtos_mais_menos_vendidos(loja_3, "Loja 3")
produtos_mais_menos_vendidos(loja_4, "Loja 4")
```

```python
import pandas as pd
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt

# Carregar os dados
urls = {
    "Loja 1": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv",
    "Loja 2": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv",
    "Loja 3": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv",
    "Loja 4": "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"
}

# Ler e combinar os dados com o nome da loja
df_total = pd.DataFrame()
for nome_loja, url in urls.items():
    df_loja = pd.read_csv(url)
    df_loja['Loja'] = nome_loja
    df_total = pd.concat([df_total, df_loja], ignore_index=True)

# ---------------------------
#  Gráfico de Dispersão Mapa Interativo (Plotly)
# ---------------------------
fig1 = px.scatter_geo(
    df_total,
    lat='lat',
    lon='lon',
    color='Loja',
    hover_name='Produto',
    scope='south america',
    title='Distribuição Geográfica das Vendas por Loja'
)
fig1.update_layout(height=600)
fig1.show()

# ---------------------------
#  Heatmap por Loja (Seaborn)
# ---------------------------
plt.figure(figsize=(10, 8))
sns.kdeplot(
    data=df_total,
    x='lon', y='lat',
    hue='Loja',
    fill=True,
    common_norm=False,
    alpha=0.6,
    palette='Set2'
)
plt.title('Heatmap - Concentração Geográfica de Vendas por Loja')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True)
plt.show()

# ---------------------------
# 📊 Desempenho por Região
# ---------------------------
desempenho = df_total.groupby(['Local da compra', 'Loja']).size().reset_index(name='Qtd Vendas')
pivot = desempenho.pivot(index='Local da compra', columns='Loja', values='Qtd Vendas').fillna(0).astype(int)
print("📌 Desempenho por Região:\n")
print(pivot)

# ---------------------------
#  Observações automáticas (Exemplo simples)
# ---------------------------
mais_vendida = df_total.groupby('Loja').size().sort_values(ascending=False)
print("\n🏆 Loja com mais vendas no total:", mais_vendida.idxmax(), f"({mais_vendida.max()} vendas)")

top_regiao = df_total['Local da compra'].value_counts().idxmax()
print("📍 Região com maior concentração de vendas:", top_regiao)
```
